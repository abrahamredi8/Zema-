```
from datasets import load_dataset
print('Loading a small Amharic FLEURS sample...')
ds = load_dataset('google/fleurs', 'am_et', split='train', streaming=True)
sample = next(iter(ds))
print('Transcript:', sample.get('transcription', 'not provided'))
print('Audio sample rate:', sample['audio']['sampling_rate'])
print('Audio samples:', len(sample['audio']['array']))
print('License: CC BY 4.0 — attribution required')
```
